```python
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

df_a  = pd.read_csv(r"D:\MAIN DATA\Documents\Semester 6\KP BRIN\Dataset\Pekerjaan\Processed\jobs_unified_summarized.csv")
df_b = pd.read_csv(r"D:\MAIN DATA\Documents\Semester 6\KP BRIN\Dataset\Pekerjaan\Processed\jobs_unified.csv")

# Model embedding lokal
model = SentenceTransformer("all-MiniLM-L6-v2")

# Ambil text
summaries = df_a["description_summary"].fillna("").astype(str).tolist()
descriptions = df_b["description"].fillna("").astype(str).tolist()

# Embedding
emb_summary = model.encode(
    summaries,
    convert_to_numpy=True,
    show_progress_bar=True
)

emb_desc = model.encode(
    descriptions,
    convert_to_numpy=True,
    show_progress_bar=True
)

# Cosine similarity per baris
similarities = np.array([
    cosine_similarity(
        emb_summary[i].reshape(1, -1),
        emb_desc[i].reshape(1, -1)
    )[0][0]
    for i in range(len(emb_summary))
])

result = pd.DataFrame({
    "summary": summaries,
    "description": descriptions,
    "similarity": similarities
})

print(result["similarity"].describe())
```

```python
print(f"Average similarity: {similarities.mean():.4f}")
print(f"Min similarity: {similarities.min():.4f}")
print(f"Max similarity: {similarities.max():.4f}")
```

```python
# Cari row similarity paling rendah
lowest = result.loc[result["similarity"].idxmin()]

print("=== PALING RENDAH ===")
print(f"Similarity: {lowest['similarity']:.4f}")
print("\nSUMMARY:")
print(lowest["summary"])
print("\nDESCRIPTION:")
print(lowest["description"])
```

```python
# Cari row similarity paling tinggi
highest = result.loc[result["similarity"].idxmax()]

print("=== PALING TINGGI ===")
print(f"Similarity: {highest['similarity']:.4f}")
print("\nSUMMARY:")
print(highest["summary"])
print("\nDESCRIPTION:")
print(highest["description"])
```
